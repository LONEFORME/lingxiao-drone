```python
#!/usr/bin/env python3
"""T265 姿态数据读取 - 显示位置坐标和朝向"""

import pyrealsense2 as rs
import time

# 初始化 pipeline
pipe = rs.pipeline()
cfg = rs.config()
cfg.enable_stream(rs.stream.pose)  # T265 的姿态流

try:
    pipe.start(cfg)
    print("T265 已启动，等待数据...\n")
    
    while True:
        frames = pipe.wait_for_frames()
        pose_frame = frames.get_pose_frame()
        
        if pose_frame:
            pose = pose_frame.get_pose_data()
            
            # 位置坐标 (单位: 米)
            tx = pose.translation.x
            ty = pose.translation.y
            tz = pose.translation.z
            
            # 速度 (米/秒)
            vx = pose.velocity.x
            vy = pose.velocity.y
            vz = pose.velocity.z
            
            # 四元数朝向
            qw = pose.rotation.w
            qx = pose.rotation.x
            qy = pose.rotation.y
            qz = pose.rotation.z
            
            # 置信度 (0-3, 3=高)
            confidence = pose.tracker_confidence
            
            print(f"\r位置: x={tx:7.3f} y={ty:7.3f} z={tz:7.3f} (m) | "
                  f"速度: vx={vx:6.3f} vy={vy:6.3f} vz={vz:6.3f} (m/s) | "
                  f"四元数: w={qw:.3f} x={qx:.3f} y={qy:.3f} z={qz:.3f} | "
                  f"置信度: {confidence}/3", end="")
        
        time.sleep(0.01)

except KeyboardInterrupt:
    print("\n\n停止...")
finally:
    pipe.stop()
```
